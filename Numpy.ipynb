```python
np.arange(1,13) # produces a range of values from 1 to 12(n-1)
```

```python
np.arange(1,13).reshape(3,4) # produces a range of values from 1 to 12 and reshapes them into an array(3-rows and 4-columns)
```

```python
np.arange(1,13).reshape(5,4) # invalid as the size of values is 12 but the array size (5*4 = 20) exceeds it
```

```python
np.arange(1,13).reshape(6,2) # size of the array 6*2=12 exactly matches with the size of the values - 12
```

```python
np.arange(1,13).reshape(6,3) # 12 doesnot equals to 6*3 = 18.. so invalid
```

```python
np.arange(1,13).reshape(12,1) # 12 equals to 12*1 = 12
```

```python
np.arange(1,13).reshape(4,3) # 12 is equal to 4*3 = 12
```

```python
np.arange(1,13).reshape(4,2) # 12 doesnot equals to 4*2 = 8.. so invalid
```
