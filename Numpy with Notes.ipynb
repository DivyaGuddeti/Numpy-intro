```python
np.arange(-16,10)
```

```python
np.arange(10,30,5) # returns a range of values from start_index till end_index-1 with a step count of 5 (10-start, 30-end, 5-step count)
```

```python
np.arange(2,9,2) # (2 - start index, 9 - end index and 2 - step count )
```

```python
np.arange(1,10,2,3) # invalid as the arange() function takes max 3 args(start,stop and step)
```

## randint() function

```python
np.random.randint(5) # produces a random value between 0 to 5 excluding 5
```

```python
np.random.randint(5,9) # produces a random value between 5 to 9 excluding 9
```

```python
np.random.randint(4,6) # returns a random number between 4 and 5.. 6 will be excluded as this will consider n-1
```

### One dimensional array

```python
np.random.randint(2,20,5) # produces 5 values between 2 to 20
```

```python
np.random.randint(2,20,4)
```

```python
np.random.randint(10,20,5)
```

### n dimensional array

```python
np.random.randint(10,40,(10,5)) # produces an array(10x5) of values from 10 till 40(excluding)
```

```python
np.arange(1,13) # produces a range of values from 1 to 12(n-1)
```

```python
np.arange(1,13).reshape(3,4) # produces a range of values from 1 to 12 and reshapes them into an array(3-rows and 4-columns)
```

```python
np.arange(1,13).reshape(5,4) # invalid as the size of values is 12 but the array size (5*4 = 20) exceeds it
```

```python
np.arange(1,13).reshape(6,2) # size of the array 6*2=12 exactly matches with the size of the values - 12
```

```python
np.arange(1,13).reshape(6,3) # 12 doesnot equals to 6*3 = 18.. so invalid
```

```python
np.arange(1,13).reshape(12,1) # 12 equals to 12*1 = 12
```

```python
np.arange(1,13).reshape(4,3) # 12 is equal to 4*3 = 12
```

```python
np.arange(1,13).reshape(4,2) # 12 doesnot equals to 4*2 = 8.. so invalid
```

#### Slicing in Matrix

```python
b = np.random.randint(10,20,(5,4)) # returns an 5x4 array of random values from 10 to 20(excludes 20)
b
```

```python
type(b) # nd arrray refers to n-dimensional array
```

```python
b
```

```python
b[:] # produces the entire array
```

```python
b[1:3] # slices the array into mentioned size(1 and 2 rows will be printed)
```

```python
b[1:3,2:3]
```
